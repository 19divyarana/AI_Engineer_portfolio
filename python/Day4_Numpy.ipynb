## NumPy = Numerical Python. It's the foundation of Data Science, Machine Learning, Deep Learning, Pandas, TensorFlow, and PyTorch.

### Python lists are flexible but slow for numerical computations.
NumPy provides:

- Faster calculations
- Multi-dimensional arrays
- Mathematical operations
- Matrix operations
- Statistical functions

```python
import numpy as np
```

```python
arr=np.array([1,2,3])
print(arr)
```

```python
arr1=np.array([[1,2,3,4],[5,6,7,8]])
print(arr1)
```

```python
print(arr1.shape)
print(arr1.ndim)
print(arr1.size)
print(arr1.dtype)
```

```python
np.zeros((3, 3))
```

```python
np.ones((2, 2))
```

```python
np.eye(3)
```

```python
np.arange(1, 11)
```

```python
arr = np.arange(1, 13)

print(arr.reshape(3, 4))
```

```python
arr = np.array([10, 20, 30, 40])

print(arr[0])
print(arr[2])
```

```python
arr = np.array([1,2,3,4,5,6])

print(arr[1:4])
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
arr = np.array([1,2,3,4,5])

print(np.sum(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
```

```python
np.random.rand(3)
```

```python
np.random.randint(1, 10, 5)
```

```python
arr = np.array([10,20,60,80,45,90])

print(arr[arr > 50])
```
