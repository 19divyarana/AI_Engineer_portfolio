```python
class Student: #class and object
    def __init__(self):
        self.name = ""

    def display(self):
        print(self.name)

s = Student()
s.name = "Divya"
s.display()
```

```python
class Student: #constructor
    def __init__(self, name):
        self.name = name

s = Student("Divya")
print(s.name)
```

```python
class Animal: #inheritencce
    def sound(self):
        print("Animal Sound")

class Dog(Animal):
    def bark(self):
        print("Barking")

d = Dog()
d.sound()
d.bark()
```

```python
class Animal: #method overriding
    def sound(self):
        print("Animal Sound")

class Dog(Animal):
    def sound(self):
        print("Dog Barks")

d = Dog()
d.sound()
```

```python
class Student: #encapsulation
    def __init__(self):
        self.__name = ""

    def set_name(self, name):
        self.__name = name

    def get_name(self):
        return self.__name

s = Student()
s.set_name("Divya")
print(s.get_name())
```

```python
class Animal {
    void sound() {
        System.out.println("Animal Sound");
    }
}

class Dog extends Animal {
    void sound() {
        System.out.println("Dog Barks");
    }
}

class Cat extends Animal {
    void sound() {
        System.out.println("Cat Meows");
    }
}
```
