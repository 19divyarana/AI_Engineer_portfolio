```python
#Data cleaning
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"],
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
}
```

```python
data = pd.DataFrame(data)
```

```python
data
```

```python
data.isnull().sum()
```

```python
data.dropna()
```

```python
data.dropna(how = "any")
```

```python
data.dropna(how = "all")
```

```python
data["Age"].fillna(data["Age"].mean())
```

```python
data["Salary"].fillna(data["Salary"].median())
```

```python
data["Age"].fillna(method = "bfill") #backward fill/ forward fill
```

```python
data.dup = data[data.duplicated(keep = "last")]
data.dup
```

```python
data = data.drop_duplicates()
data
```

```python
data["Promoted Salary"] = data["Salary"] * 10
data
```

```python
data["Promoted Salary"] = data["Promoted Salary"].apply(lambda x: x/10 if x > 650000 else x)
data
```

```python
# apply and lambda
def multiplying_age(x):
  return x*2

data["Age"] = data["Age"].apply(multiplying_age)
```

```python
data
```

```python
data["Age"] = data["Age"].apply(lambda x: x/2)
```
