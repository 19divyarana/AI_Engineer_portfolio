```python
import numpy as np
arr = np.array([5,2,9,1,3]) #sorting

print(np.sort(arr))
```

```python
arr = np.array([10,20,30,40,50]) #filtering

print(arr[arr > 30])
```

```python
arr = np.array([1,2,3,4,5]) #boolean masking

mask = arr % 2 == 0

print(mask)
print(arr[mask])
```

```python
arr = np.array([ #broadcasting
    [1,2,3],
    [4,5,6]
])

print(arr + 10)
```

```python
A = np.array([ #matrix operations
    [1,2],
    [3,4]
])

B = np.array([
    [5,6],
    [7,8]
])

print(np.dot(A, B))
```

```python
A = np.array([ #transpose
    [1,2,3],
    [4,5,6]
])

print(A.T)
```

```python
A = np.array([   #inverse matrix
    [1,2],
    [3,4]
])

print(np.linalg.inv(A))
```

```python
A = np.array([  #det
    [1,2],
    [3,4]
])

print(np.linalg.det(A))
```

```python
a = np.array([1,2,3]) #concat
b = np.array([4,5,6])

print(np.concatenate((a, b)))
```

```python
a = np.array([[1,2]])
b = np.array([[3,4]])

print(np.vstack((a,b)))
```

```python
a = np.array([[1,2]])
b = np.array([[3,4]])

print(np.hstack((a,b)))
```

```python
arr = np.array([1,2,2,3,4,4])

print(np.unique(arr))
```

```python
arr = np.array([10,20,30,40])

print(np.where(arr[] > 20))
```
