```python
"""Create a Student class with:

name
age

Add a method display() to print student details."""

class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def display(self):
            print(f"Name :{self.name}")
            print(f"Age : {self.age}")
s=Student("Div",21)
s.display()
```

```python
"""Create a Rectangle class with:

length
breadth

Add method area()."""

class Rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth

    def area(self):
        return self.length*self.breadth

r=Rectangle(10,5)
print(r.area())
```

```python
"""Create a BankAccount class with:

balance

Methods:

deposit()
withdraw()
show_balance()"""

class BankAccount:
    def __init__(self,balance=0):
        self.balance=balance
    def deposit(self,amount):
        self.balance +=amount
    def withdraw(self, amount):
        self.balance -= amount
    def show_balance(self):
        print("Balance",self.balance)
b=BankAccount(23000)
b.deposit(500)
b.withdraw(200)
b.show_balance()
```

```python
"""Create:

Parent class: Animal
Child class: Dog

Both should have methods."""
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

d = Dog()
d.eat()
d.bark()
```

```python
"""Create:

Car
Bike

Both should have a move() method."""
class Car:
    def move(self):
        print("Car is moving")

class Bike:
    def move(self):
        print("Bike is moving")

vehicles = [Car(), Bike()]

for v in vehicles:
    v.move()
```

```python
"""Create abstract class Shape with area() method.
Answer"""
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

c = Circle(5)
print(c.area())
```
